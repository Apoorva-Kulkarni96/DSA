### Two Sum : Check if a pair with given sum exists in Array

#### Brute force Approach

```python
def TwoSumExits(arr,k):
    n = len(arr)

    for i in range(n):
        for j in range(n):
            if arr[i] + arr[j] == k:
                return "YES"
    return "NO"

def TwoSumIndicies(arr,k):
    n = len(arr)

    for i in range(n):
        for j in range(n):
            if arr[i] + arr[j] == k:
                return i,j
    return (-1,-1)
```

Time Complexity: O(n^2)

Space Complexity: O(1)

#### Better Approach


```python
def TwoSumExits(arr,k):
    n = len(arr)
    mp ={}
    complement = 0
    for i, num in enumerate(arr):
        complement = k - num
        if complement in mp:
            return "Yes"
        else:
            mp[num] = i
    return "No"

def TwoSumIndices(arr,k):
    n = len(arr)
    mp ={}
    complement = 0
    for i,num in enumerate(arr):
        complement = k - num
        if complement in mp:
            return (mp[complement], i)
        else:
            mp[num] = i
    return -1,-1
```

Time Complexity: O(n)

Space Complexity: O(n)

#### Optimal Approach

```python
def TwoPointerSumExits(arr,k):
    n = len(arr)
    left = 0
    right = n - 1
    curr_sum = 0
    arr.sort()
    while left < right:
        curr_sum = arr[left] + arr[right]
        if curr_sum == k:
            return "YES"
        elif curr_sum > k:
            right -= 1
        else:
            left += 1

    return "NO"

def TwoPointerSumIndices(arr,k):
    n = len(arr)
    left = 0
    right = n-1
    curr_sum = 0
    arr_sort_index = [(num, i) for i, num in enumerate(arr)]
    arr_sort_index.sort()
    while left < right:
        curr_sum = arr_sort_index[left][0] + arr_sort_index[right][0]
        if curr_sum == k:
            return arr_sort_index[left][1], arr_sort_index[right][1]
        elif curr_sum > k:
            right -= 1
        else:
            left += 1
    return (-1,-1)
```

Time Complexity: O(n)

Space Complexity: O(1)

### Sort an array of 0's 1's and 2's

#### Brute Approach



```python
def sortzerosones(arr):
    n = len(arr)
    count0=count1=count2=0
    for num in arr:
        if num == 0:
            count0 += 1
        elif num == 1:
            count1 += 1
        elif num == 2:
            count2 += 1
    index = 0
    for _ in range(count0):
        arr[index] = 0
        index += 1
    for _ in range(count1):
        arr[index] = 1
        index += 1
    for _ in range(count2):
        arr[index] = 2
        index += 1
    return arr
```

```python
sortzerosones([1, 0, 2, 1, 0])
```

Time Complexity: O(n)

Space Complexity: O(1)

#### Better Approach

```python
def SortZeroOnes(arr):
    n = len(arr)
    count0 = count1 = count2 = 0
    for num in arr:
        if num == 0:
            count0 += 1
        if num == 1:
            count1 += 1
        if num == 1:
            count2 += 1
    
    for i in range(count0):
        arr[i] = 0
    for i in range(count0,count0+count1):
        arr[i] = 1
    for i in range(count0+count1, n):
        arr[i] = 2
    return arr
```

```python
SortZeroOnes([1, 0, 2, 1, 0])
```
