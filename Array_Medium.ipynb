### Two Sum : Check if a pair with given sum exists in Array

#### Brute force Approach

```python
def TwoSumExits(arr,k):
    n = len(arr)

    for i in range(n):
        for j in range(n):
            if arr[i] + arr[j] == k:
                return "YES"
    return "NO"

def TwoSumIndicies(arr,k):
    n = len(arr)

    for i in range(n):
        for j in range(n):
            if arr[i] + arr[j] == k:
                return i,j
    return (-1,-1)
```

Time Complexity: O(n^2)

Space Complexity: O(1)

#### Better Approach


```python
def TwoSumExits(arr,k):
    n = len(arr)
    mp ={}
    complement = 0
    for i, num in enumerate(arr):
        complement = k - num
        if complement in mp:
            return "Yes"
        else:
            mp[num] = i
    return "No"

def TwoSumIndices(arr,k):
    n = len(arr)
    mp ={}
    complement = 0
    for i,num in enumerate(arr):
        complement = k - num
        if complement in mp:
            return (mp[complement], i)
        else:
            mp[num] = i
    return -1,-1
```

Time Complexity: O(n)

Space Complexity: O(n)

#### Optimal Approach

```python
def TwoPointerSumExits(arr,k):
    n = len(arr)
    left = 0
    right = n - 1
    curr_sum = 0
    arr.sort()
    while left < right:
        curr_sum = arr[left] + arr[right]
        if curr_sum == k:
            return "YES"
        elif curr_sum > k:
            right -= 1
        else:
            left += 1

    return "NO"

def TwoPointerSumIndices(arr,k):
    n = len(arr)
    left = 0
    right = n-1
    curr_sum = 0
    arr_sort_index = [(num, i) for i, num in enumerate(arr)]
    arr_sort_index.sort()
    while left < right:
        curr_sum = arr_sort_index[left][0] + arr_sort_index[right][0]
        if curr_sum == k:
            return arr_sort_index[left][1], arr_sort_index[right][1]
        elif curr_sum > k:
            right -= 1
        else:
            left += 1
    return (-1,-1)
```

Time Complexity: O(n)

Space Complexity: O(1)

### Sort an array of 0's 1's and 2's

#### Brute Approach



```python
def sortzerosones(arr):
    n = len(arr)
    count0=count1=count2=0
    for num in arr:
        if num == 0:
            count0 += 1
        elif num == 1:
            count1 += 1
        elif num == 2:
            count2 += 1
    index = 0
    for _ in range(count0):
        arr[index] = 0
        index += 1
    for _ in range(count1):
        arr[index] = 1
        index += 1
    for _ in range(count2):
        arr[index] = 2
        index += 1
    return arr
```

```python
sortzerosones([1, 0, 2, 1, 0])
```

Time Complexity: O(n)

Space Complexity: O(1)

#### Better Approach

```python
def SortZeroOnes(arr):
    n = len(arr)
    count0 = count1 = count2 = 0
    for num in arr:
        if num == 0:
            count0 += 1
        if num == 1:
            count1 += 1
        if num == 1:
            count2 += 1
    
    for i in range(count0):
        arr[i] = 0
    for i in range(count0,count0+count1):
        arr[i] = 1
    for i in range(count0+count1, n):
        arr[i] = 2
    return arr
```

```python
SortZeroOnes([1, 0, 2, 1, 0])
```

Time Complexity: O(n)

Space Complexity: O(1)

#### Optimal Approach(Dutch National Flag)

```python
def SortZeroOnes(arr):
    n = len(arr)
    low = 0
    mid = 0
    high = n - 1
    while mid <= high:
        if arr[mid] == 0:
            arr[mid], arr[low] = arr[low], arr[mid]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        else:
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
    return arr

    
```

```python
SortZeroOnes([1, 0, 2, 1, 0 ])
```

Time Complexity: O(n)

Space Complexity: O(1)

### Find the Majority Element that occurs more than N/2 times
 nums = [7, 0, 0, 1, 7, 7, 2, 7, 7]

```python
def MajorityEle(arr):
    n = len(arr)
    for i in range(n):
        cnt=0
        for j in range(n):
            if arr[i] == arr[j]:
                cnt += 1
        if cnt > n//2:
            return arr[i]
    return -1
```

```python
MajorityEle([7, 0, 0, 1, 7, 7, 2, 7, 7])
```

Time Complexity: O(n^2)

Space Complexity: O(1)

#### Better Approach

```python
def MajorityEle(arr):
    n = len(arr)
    seen = {}

    for num in arr:
        if num in seen:
            seen[num] += 1
        else:
            seen[num] = 1
    
    for i,cnt in seen.items():
        if cnt > n//2:
            return i
    return -1
```

```python
MajorityEle([7, 0, 0, 1, 7, 7, 2, 7, 7])
```

Time Complexity: O(n)

Space Complexity: O(n)

#### Better Approach


```python
def MajorityEle(arr):
    n = len(arr)
    cnt = 0
    ele = 0
    for num in arr:
        if cnt == 0:
            ele = num
            cnt = 1
        elif ele == num:
            cnt += 1
        else:
            cnt -= 1

    cnt1 = arr.count(ele)

    if cnt1 > n//2:
        return ele
    return -1
```

```python
MajorityEle([7, 0, 0, 1, 7, 7, 2, 7, 7])
```

Time Complexity: O(n)

Space Complexity: O(1)

### Kadane's Algorithm : Maximum Subarray Sum in an Array
[2, 3, 5, -2, 7, -4]  
Output:
 15  
Explanation:
 The subarray from index 0 to index 4 has the largest sum = 15

#### Brute Approach

```python
def Kadane(arr):
    n = len(arr)
    maxi = float('-inf')

    for startnum in range(n):
        curr_sum = 0
        for endnum in range(startnum, n):
            curr_sum += arr[endnum]
            maxi = max(maxi, curr_sum)

    return maxi
```

```python
Kadane([2, 3, 5, -2, 7, -4])
```

Time Complexity: O(n^2)

Space Complexity: O(1)

#### Optimal Solution

```python
def Kadane(arr):
    curr_sum = 0
    maxi = float('-inf')
    for i in range(len(arr)):
        curr_sum += arr[i]

        if curr_sum > maxi:
            maxi = curr_sum
        if curr_sum < 0:
            curr_sum = 0
    return maxi
```

```python
Kadane([2, 3, 5, -2, 7, -4])
```

Time Complexity: O(n^2)

Space Complexity: O(1)

### Best Time to Buy and Sell Stock

#### Brute Force Approach

```python
def stock(arr):
    n = len(arr)
    max_profit = profit = 0
    for buy in arr:
        for sell in range(buy, n):
            profit = arr[sell] - arr[buy]
            max_profit = max(max_profit, profit)
    return max_profit

```

```python
stock([7,1,5,3,6,4])
```

Time Complexity: O(n^2)

Space Complexity: O(1)

#### Optimal Solution

```python
def stock(arr):
    min_price = float('inf')
    max_profit = profit = 0
    for price in arr:
        if price < min_price:
            min_price = price
        else:
            profit = price - min_price
            max_profit = max(profit, max_profit)
    return max_profit

        
```

```python
stock([7,1,5,3,6,4])
```

Time Complexity: O(n)

Space Complexity: O(1)

### Rearrange Array Elements by Sign

### Brute Approach

```python
def ArrayManipulator(arr):
    pos_num = []
    neg_num = []
    n = len(arr)
    for i in range(n):
        if arr[i] > 0:
            pos_num.append(arr[i])
        else:
            neg_num.append(arr[i])
    for i in range(n // 2):
        arr[2 * i] = pos_num[i]
        arr[2 * i + 1] = neg_num[i] 

    return arr
            
```

```python
ArrayManipulator([1, 2, -4, -5])
```

Time Complexity: O(n)

Space Complexity: O(n)

### next_permutation : find next lexicographically greater permutation

#### Brute Force

Input: Arr[] = {1,3,2}
Output: {2,1,3}
Explanation: All permutations of {1,2,3} are {{1,2,3} , {1,3,2}, {2,13} , {2,3,1} , {3,1,2} , {3,2,1}}. So, the next permutation just after {1,3,2} is {2,1,3}.


```python
from itertools import permutations

def next_permutations(arr):
    nums = sorted(set(permutations(arr)))
    n = len(nums)
    current = tuple(arr)
    for i in range(n):
        if i == n-1:
            return list(nums[0])
        if current == nums[i]:
            return list(nums[i+1])
    
        
        
```

```python
next_permutations([3,2,1])
```
